# PydanticAI fundamentals

```python
from pydantic_ai import Agent

# GOOGLE_API_KEY or GEMINI_API_KEY must be set in your .env file or environment variables
agent = Agent(model="google-gla:gemini-2.5-flash")

result = await agent.run("Give me a programming joke")

result
```

```python
print(result.output)
```

```python
result = await agent.run("Give me an IT employee working in Sweden, shortly describe their role, and suggest a suitable programming language for them to learn.")
result
```

```python
print(result.output)
```

```python
with open("employee.md", "w") as file:
    file.write(result.output)
```

## pydantic to structure the output

```python
from pydantic import BaseModel, Field

class EmployeeModel(BaseModel):
    name: str = Field(description="The full name of the employee")
    age: int = Field(description="The age of the employee in years", gt = 17, lt = 70)
    salary: int = Field(description="The monthly salary of the employee between 3500 and 70000", gt=35000, lt=70000)
    role: str = Field(description="The job role of the employee")

result = await agent.run("Give me an IT employee working in Sweden, shortly describe their role, and suggest a suitable programming language for them to learn.", output_type=EmployeeModel)
result
```

```python
employee = result.output
employee
```

```python
isinstance(employee, EmployeeModel), isinstance(employee, BaseModel)
```
