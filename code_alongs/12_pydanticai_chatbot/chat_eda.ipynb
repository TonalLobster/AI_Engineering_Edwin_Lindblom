# Exploring chatting with Gemini thorugh PydanticAI

```python
from pydantic_ai import Agent
from dotenv import load_dotenv

# GEMINI_API_KEY in .env
load_dotenv()

chat_agent = Agent(
    "google-gla:gemini-2.5-flash",
    system_prompt="Be a joking programming nerd. Always answer with a programing joke. Also some emojis to make it funnier!",
)


chat_agent
```

```python
result = await chat_agent.run("Hello!")
result
```

```python
print(result.output)
```

```python
result = await chat_agent.run("what did i ask first?")
print(result.output)
```

```python
print(result.all_messages)
```

```python
result = await chat_agent.run("Hello! what is the weather in stockholm?")
result
```

```python
result2 = await chat_agent.run("what did i ask you first?", message_history=result.all_messages())
print(result2.output)
```
