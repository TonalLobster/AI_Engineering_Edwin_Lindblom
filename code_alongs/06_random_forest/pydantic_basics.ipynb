## validate our Person class

```python
class Person:
    def __init__(self, name:str, gender:str, age:int):
        if not isinstance(name, str):
            raise TypeError(f"{name}, must be of type str, not {type(name)}")
        self.name = name
        self.gender = gender
        self.age = age


# getter
@property
def age(self):
    return self._age

@age.setter
def age( self, value: int):
    if not isinstance(value,int):
        raise TypeError(f"age must be of type int, not {type(age)} that you have provided")
    if value < 0 or value > 125:
        raise ValueError(f"age must be between 0 and 125, not {value} that you provided")
    self._age = value


person5 = Person(name = "Bella", gender = "F", age = 3)
person5.age,person5.gender, person5.name
```

```python
person5 = Person(name = "Bella", gender = "F", age = -3)
```

## Work Smarter with pydantic

```python
from pydantic import BaseModel

# inherits from BaseModel -> makes it into pydantic model, but still a normal python class
class Person(BaseModel):
    name: str
    gender: str
    age: int

person6 = Person(name= "Tjorven", gender = "M", age = 35)
person6
```

```python
person6.name, person6.age, person6.gender
```

```python
person6.age = 36
person6.age
```

```python
person6
```

```python
from pydantic import ValidationError
try:
    Person(name = 3.1415, gender = 2.714, age = -5)
except ValidationError as err:
    print(err)
```

#### bad we have negative age

```python
Person(name = "Filip", gender="M", age = -5)
```

### Fix Person class with age validation

```python
from pydantic import Field

class Person(BaseModel):
    name: str
    gender: str
    age: int = Field(gt=1, lt=126)

try:
    Person(name = "Edvin", gender = "M", age = 173)
except ValidationError as err:
    print(err)
```

```python
from typing import Literal

class Person(BaseModel):
    name: str
    gender: Literal["M", "F"]
    age: int = Field(gt=1, lt=126)

try:
    Person(name = "Edvin", gender = "Female", age = 26)
except ValidationError as err:
    print(err)


Person(name = "Edvin", gender = "F", age = 26)
```
