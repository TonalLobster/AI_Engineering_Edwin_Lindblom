## PydanticAI fundamentals

```python
from pydantic_ai import Agent
from pydantic import BaseModel, Field

class Restaurant(BaseModel):
    name: str
    cuisine: str
    price_level: str
    rating: int = Field(gt=0, le=10)
    short_description: str
    opening_hours: str
    location: str

class RestaurantList(BaseModel):
    restaurants: list[Restaurant]


agent = Agent(
    model="google-gla:gemini-2.5-flash",
    output_type=RestaurantList,
    system_prompt= """
    Your are a restaurant recommendation agent. Based on the location provided by the user, recommend a list of 5 restaurants. Each restaurant should have the following details:
    - name
    - cuisine
    - price_level (e.g., $, $$, $$$)
    - rating (an integer between 1 and 10)
    - short_description
    - opening_hours
    - location (address or area)
    """
    )

result = await agent.run("Hi! Can you recommend some good restaurants in Gothenburg,Sweden? :)")
```

```python
import pandas as pd

restaurants_list = [vars(r) for r in result.output.restaurants]
df = pd.DataFrame(restaurants_list)
print(df)
```
